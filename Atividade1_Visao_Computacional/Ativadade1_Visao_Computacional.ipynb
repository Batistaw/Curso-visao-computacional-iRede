**Transformações de Imagens**

**Perda de detalhe percebida**
* Ao reduzir a imagem para 50%, nota-se uma leve diminuição na nitidez das bordas. No entanto, na redução para 20%, a perda de informações é drástica.

**Possível impacto em Visão Computacional**
* Uma redução muito altaa na resolução, comprometeria seriamente o desempenho de algoritmos focados em extração de característica de borda ou reconhecimentos de texturas.




```python
## Exemplo da classe 'Mouse'
img_mouse_original = cv2.imread('/content/dataset/classe_Mouse/img006_original.jpg') # carregando a imagem original
## Versão com redução em 50%
# Obtém a largura e altura originais da imagem
largura_mouse, altura_mouse = img_mouse_original.shape[:2]
largura_mouse_reduzida_50 = largura_mouse // 2 # Resolução da largura reduzia em 50%
altura_mouse_reduzida_50 = altura_mouse // 2 # Resolução da altura reduzida em 50%
# Reduzindo a imagem em 50%
imagem_mouse_resolucao_reduzida_50 = cv2.resize(img_mouse_original, (largura_mouse_reduzida_50, altura_mouse_reduzida_50), interpolation=cv2.INTER_AREA)

## Versão com redução em 20%
largura_mouse_reduzida_20 = largura_mouse // 5 # Resolução da largura reduzia em 50%
altura_mouse_reduzida_20 = altura_mouse // 5 # Resolução da altura reduzida em 50%
# Reduzindo a imagem em 20%
imagem_mouse_resolucao_reduzida_20 = cv2.resize(img_mouse_original, (largura_mouse_reduzida_20, altura_mouse_reduzida_20))

print("Mouse Original: ", img_mouse_original.shape)
print("Mouse Redução de 50%: ", imagem_mouse_resolucao_reduzida_50.shape)
print("Mouse Redução de 20%: ", imagem_mouse_resolucao_reduzida_20.shape)

## Comparação das imagens da classe 'Mouse'
fig, eixos = plt.subplots(1, 3, figsize=(15, 4)) # Cria três áres lado a lado para comparação
eixos[0].imshow(img_mouse_original)
eixos[0].set_title('Original')
eixos[0].axis("off")  # Remove os eixos da imagem
eixos[1].imshow(imagem_mouse_resolucao_reduzida_50)
eixos[1].set_title('Redução de 50%')
eixos[1].axis("off")  # Remove os eixos da imagem
eixos[2].imshow(imagem_mouse_resolucao_reduzida_20)
eixos[2].set_title('Redução de 20%')
eixos[2].axis("off")  # Remove os eixos da imagem


## Exemplo da classe 'Teclado'
# Versão original da imagem
img_teclado_original = cv2.imread('/content/dataset/classe_Teclado/img001_original.jpg')
## Versão com redução em 50%
# Obtém a largura e altura originais da imagem
largura_teclado, altura_teclado = img_teclado_original.shape[:2]
largura_teclado_reduzida_50 = largura_teclado // 2 # Resolução da largura reduzia em 50%
altura_teclado_reduzida_50 = altura_teclado // 2 # Resolução da largura reduzia em 50%
# Reduzindo a imagem em 50%
imagem_teclado_resolucao_reduzida_50 = cv2.resize(img_teclado_original, (largura_teclado_reduzida_50, altura_teclado_reduzida_50))

## Versão reduzida em 20%
lagura_teclado, altura_teclado = img_teclado_original.shape[:2]
largura_teclado_reduzida_20 = largura_teclado // 5 # Resolução da largura reduzia em 20%
altura_teclado_reduzida_20 = altura_teclado // 5 # Resolução da largura reduzia em 20%
# Reduzindo a imagem em 20%
imagem_teclado_resolucao_reduzida_20 = cv2.resize(img_teclado_original, (largura_teclado_reduzida_20, altura_teclado_reduzida_20))

print("\nTeclado Original: ", img_teclado_original.shape) # Plota a resolução da imagem Original
print("Teclado Redução de 50%: ", imagem_teclado_resolucao_reduzida_50.shape) # Plota a resolução da imagem reduzida em 50%
print("Teclado Redução de 20%: ", imagem_teclado_resolucao_reduzida_20.shape) # Plota a resolução da imagem reduzida em 20%

## Comparação das imagens da classe 'Mouse'
fig, eixos = plt.subplots(1, 3, figsize=(15, 4)) # Cria três áres lado a lado para comparação
eixos[0].imshow(img_teclado_original)
eixos[0].set_title('Original')
eixos[0].axis("off")  # Remove os eixos da imagem
eixos[1].imshow(imagem_teclado_resolucao_reduzida_50)
eixos[1].set_title('Redução de 50%')
eixos[1].axis("off")  # Remove os eixos da imagem
eixos[2].imshow(imagem_teclado_resolucao_reduzida_20)
eixos[2].set_title('Redução de 20%')
eixos[2].axis("off")  # Remove os eixos da imagem


```

**Espaço De Cor**

**Quais diferenças visuais aparecem ao converter**
* Em RGB a imagem mantem os 3 canais visuais, a representação de como o olho humano enxerga.
Escala de Cinza: A imagem perde toda a informação de cor e passa a exibir apenas 1 canal de cor e o contraste entre luz e sombra fica mais evidentes
HSV a imagem ganha um aspecto "psicodélico", pois os canais separam a cor pura, da sua pureza e da intensidade de luz

**Em que tipo de tarefa usaria-se cada cor**
* O RGB é ideal para tarefas onde a cor é a principal característica distintiva, como classificar o amadurecimento de uma maçã.
Na Escala de cinza, muito utilizada no reconhecimento facial, detecção de bordas, OCR e detecção de formas geométricas.
HSV, recomendada para rastreamente de objetos coliridos em vídeo ou segmentação baseada em cor.

```python
## Exemplo da classe Mouse
imagem_mouse_original_cor = cv2.imread('/content/dataset/classe_Mouse/img007_original.jpg')
imagem_mouse_RGB = cv2.cvtColor(imagem_mouse_original_cor, cv2.COLOR_BGR2RGB)
# converte a imagem para escala de cinza
imagem_mouse_cinza = cv2.cvtColor(imagem_mouse_RGB, cv2.COLOR_RGB2GRAY)
# Converte a imagem para HSV
imagem_mouse_HSV = cv2.cvtColor(imagem_mouse_RGB, cv2.COLOR_RGB2HSV)

fig, eixos = plt.subplots(1, 3, figsize=(15, 4)) # Cria três áres lado a lado para comparação
eixos[0].imshow(imagem_mouse_original_cor)
eixos[0].set_title('RGB')
eixos[0].axis("off")  # Remove os eixos da imagem
eixos[1].imshow(imagem_mouse_cinza, cmap='gray')
eixos[1].set_title('Cinza')
eixos[1].axis("off")  # Remove os eixos da imagem
eixos[2].imshow(imagem_mouse_HSV)
eixos[2].set_title('HSV')
eixos[2].axis("off")  # Remove os eixos da imagem


## Exemplo da classe Teclado
imagem_teclado_original_cor = cv2.imread('/content/dataset/classe_Teclado/img002_original.jpg')
imagem_teclado_RGB = cv2.cvtColor(imagem_teclado_original_cor, cv2.COLOR_BGR2RGB)
# converte a imagem para escala de cinza
imagem_teclado_cinza = cv2.cvtColor(imagem_teclado_RGB, cv2.COLOR_RGB2GRAY)
# Converte a imagem para HSV
imagem_teclado_HSV = cv2.cvtColor(imagem_teclado_RGB, cv2.COLOR_RGB2HSV)

fig, eixos = plt.subplots(1, 3, figsize=(15, 4)) # Cria três áres lado a lado para comparação
eixos[0].imshow(imagem_teclado_original_cor)
eixos[0].set_title('RGB')
eixos[0].axis("off")  # Remove os eixos da imagem
eixos[1].imshow(imagem_teclado_cinza, cmap='gray')
eixos[1].set_title('Cinza')
eixos[1].axis("off")  # Remove os eixos da imagem
eixos[2].imshow(imagem_teclado_HSV)
eixos[2].set_title('HSV')
eixos[2].axis("off")  # Remove os eixos da imagem
```

**Quantização**

**O que muda visualmente**
* Na versão com 256 níveis, as transições de luz e sombra são totalmente suaves. Ao reduzir para 64 e 32 níveis, as transições deixam de ser contínuas e começam a surgir blocos demarcados na iluminação (fenômeno conhecido como falsos contornos ou *banding*). Na versão extrema de 2 níveis, a imagem vira uma representação binária, perdendo completamente a noção de volume, profundidade e textura do mouse/teclado.

**Qual versão é mais útil para o dataset**
* A versão de 64 níveis poderia ser útil para o treinamento do classificador, pois a forma estrutural do objeto não foi perdida e a redução de informação tonal ajudaria a reduzir o processamento necessário.

```python
def quantizar(imagem, niveis):
  passo = 256 / niveis
  resultado = np.floor(imagem.astype(np.float32) / passo) * passo
  return np.clip(resultado, 0, 255).astype(np.uint8)

## Exemplo com a classe mouse
imagem_mouse_quantizacao = plt.imread('/content/dataset/classe_Mouse/img008_original.jpg')
#  Transformação da imagem original em escala de cinza
imagem_mouse_cinza_quantizaco = cv2.cvtColor(imagem_mouse_quantizacao, cv2.COLOR_BGR2GRAY)
# Mantém aproximadamente todos os níveis
imagem_mouse_256 = quantizar(imagem_mouse_cinza_quantizaco, 256)
# Reduz a imagem para 64 níveis de intensidade
imagem_mouse_64 = quantizar(imagem_mouse_cinza_quantizaco, 64)
# Reduz a imagem para 32 níveis de intensidade
imagem_mouse_32 = quantizar(imagem_mouse_cinza_quantizaco, 32)
# Reduz a imagem para 2 níveis de intensidade
imagem_mouse_2 = quantizar(imagem_mouse_cinza_quantizaco, 2)
# Plotando as imagens
fig, eixos = plt.subplots(1, 4, figsize=(16, 4))  # Cria quatro áreas para comparar diferentes níveis de quantização.
eixos[0].imshow(imagem_mouse_256, cmap="gray", vmin=0, vmax=255)  # Exibe a versão com 256 níveis.
eixos[0].set_title("256 níveis")  # Identifica a primeira quantização.
eixos[0].axis("off")  # Remove os eixos da primeira imagem.
eixos[1].imshow(imagem_mouse_64, cmap="gray", vmin=0, vmax=255)  # Exibe a versão com 64 níveis.
eixos[1].set_title("64 níveis")  # Identifica a segunda quantização.
eixos[1].axis("off")  # Remove os eixos da segunda imagem.
eixos[2].imshow(imagem_mouse_32, cmap="gray", vmin=0, vmax=255)  # Exibe a versão com 32 níveis.
eixos[2].set_title("32 níveis")  # Identifica a terceira quantização.
eixos[2].axis("off")  # Remove os eixos da terceira imagem.
eixos[3].imshow(imagem_mouse_2, cmap="gray", vmin=0, vmax=255)  # Exibe a versão com 2 níveis.
eixos[3].set_title("2 níveis")  # Identifica a quarta quantização.
eixos[3].axis("off")  # Remove os eixos da quarta imagem.
plt.tight_layout()  # Ajusta o espaçamento entre as quatro imagens.
plt.show()  # Exibe a comparação das quantizações.

## Exemplo com a classe Teclado
imagem_teclado_quantizacao = plt.imread('/content/dataset/classe_Teclado/img004_original.jpg')
#  Transformação da imagem original em escala de cinza
imagem_teclado_cinza_quantizaco = cv2.cvtColor(imagem_teclado_quantizacao, cv2.COLOR_BGR2GRAY)
# Mantém aproximadamente todos os níveis
imagem_teclado_256 = quantizar(imagem_teclado_cinza_quantizaco, 256)
# Reduz a imagem para 64 níveis de intensidade
imagem_teclado_64 = quantizar(imagem_teclado_cinza_quantizaco, 64)
# Reduz a imagem para 32 níveis de intensidade
imagem_teclado_32 = quantizar(imagem_teclado_cinza_quantizaco, 32)
# Reduz a imagem para 2 níveis de intensidade
imagem_teclado_2 = quantizar(imagem_teclado_cinza_quantizaco, 2)
# Plotando as imagens
fig, eixos = plt.subplots(1, 4, figsize=(16, 4))  # Cria quatro áreas para comparar diferentes níveis de quantização.
eixos[0].imshow(imagem_teclado_256, cmap="gray", vmin=0, vmax=255)  # Exibe a versão com 256 níveis.
eixos[0].set_title("256 níveis")  # Identifica a primeira quantização.
eixos[0].axis("off")  # Remove os eixos da primeira imagem.
eixos[1].imshow(imagem_teclado_64, cmap="gray", vmin=0, vmax=255)  # Exibe a versão com 64 níveis.
eixos[1].set_title("64 níveis")  # Identifica a segunda quantização.
eixos[1].axis("off")  # Remove os eixos da segunda imagem.
eixos[2].imshow(imagem_teclado_32, cmap="gray", vmin=0, vmax=255)  # Exibe a versão com 32 níveis.
eixos[2].set_title("32 níveis")  # Identifica a terceira quantização.
eixos[2].axis("off")  # Remove os eixos da terceira imagem.
eixos[3].imshow(imagem_teclado_2, cmap="gray", vmin=0, vmax=255)  # Exibe a versão com 2 níveis.
eixos[3].set_title("2 níveis")  # Identifica a quarta quantização.
eixos[3].axis("off")  # Remove os eixos da quarta imagem.
plt.tight_layout()  # Ajusta o espaçamento entre as quatro imagens.
plt.show()  # Exibe a comparação das quantizações.

```

**Formato de Arquivo**

**Tamanho do arquivo (KB/MB):**
* **Classe Mouse:** PNG (8,71 MB) | JPEG Qualidade 25 (208 KB).
* **Classe Teclado:** PNG (9,18 MB) | JPEG Qualidade 25 (212 KB).

**Difereçã visual Percebida**
* Enquanto o PNG preservou o valor exato de todos os pixels da imagem, o JPEG com fator de compressão elevado (25) apresentou distorções visuais evidentes. Surgiram "artefatos de bloco" ao redor das teclas e na transição suave do plástico do mouse para o fundo, criando padrões artificiais quadradinhos e desfocados que não existem na cena real.

**Hipótese sobre impacto  em tarefas de PDI**
* A alta compressão do JPEG gera "bordas falsas" e borra texturas finas. Se essas imagens comprimidas fossem enviadas para um algoritmo detector de defeitos de fabricação (em uma esteira industrial, por exemplo), os artefatos do JPEG gerariam ruídos geométricos que o algoritmo poderia classificar erroneamente como arranhões, causando falsos positivos. O PNG seria ideal para essas áreas que exigem precisão rigorosa em cada pixel, porém, o PNG exigiria um custo de armazenamento e carregamento de memória imensamente maior para treinar um modelo de Machine Learning em larga escala

```python
from PIL import Image  # Biblioteca Pillow para codificar imagens em diferentes formatos.
from io import BytesIO  # BytesIO para armazenar arquivos temporariamente na memória.

## Exemplo com a classe mouse
imagem_mouse_formataco = plt.imread('/content/dataset/classe_Mouse/img010_original.jpg')
# Conversão da imagem para RGB
imagem_mouse_formataco_RGB = cv2.cvtColor(imagem_mouse_formataco, cv2.COLOR_BGR2RGB)
buffer_png = BytesIO()  # Cria um buffer em memória para armazenar a versão PNG.
Image.fromarray(imagem_mouse_formataco_RGB).save(buffer_png, format="PNG")  # Codifica a imagem em PNG sem perdas.
buffer_jpeg = BytesIO()  # Cria um buffer em memória para armazenar a versão JPEG.
Image.fromarray(imagem_mouse_formataco_RGB).save(buffer_jpeg, format="JPEG", quality=25)  # Codifica a imagem em JPEG com compressão elevada.
tamanho_png = len(buffer_png.getvalue())  # Calcula o tamanho em bytes da representação PNG.
tamanho_jpeg = len(buffer_jpeg.getvalue())  # Calcula o tamanho em bytes da representação JPEG.
print("Mouse")
print("PNG:", tamanho_png, "bytes")  # Exibe o tamanho aproximado do arquivo PNG.
print("JPEG qualidade 25:", tamanho_jpeg, "bytes")  # Exibe o tamanho aproximado do JPEG comprimido.

## Exemplo da classe Teclado
# Conversão da imagem para
imagem_teclado_formataco = plt.imread('/content/dataset/classe_Teclado/img005_original.jpg')
imagem_teclado_formataco_RGB = cv2.cvtColor(imagem_teclado_formataco, cv2.COLOR_BGR2RGB)
buffer_png = BytesIO()  # Cria um buffer em memória para armazenar a versão PNG.
Image.fromarray(imagem_teclado_formataco_RGB).save(buffer_png, format="PNG")  # Codifica a imagem em PNG sem perdas.
buffer_jpeg = BytesIO()  # Cria um buffer em memória para armazenar a versão JPEG.
Image.fromarray(imagem_teclado_formataco_RGB).save(buffer_jpeg, format="JPEG", quality=25)  # Codifica a imagem em JPEG com compressão elevada.
tamanho_png = len(buffer_png.getvalue())  # Calcula o tamanho em bytes da representação PNG.
tamanho_jpeg = len(buffer_jpeg.getvalue())  # Calcula o tamanho em bytes da representação JPEG.
print("\nTeclado")
print("PNG:", tamanho_png, "bytes")  # Exibe o tamanho aproximado do arquivo PNG.
print("JPEG qualidade 25:", tamanho_jpeg, "bytes")  # Exibe o tamanho aproximado do JPEG comprimido.
```
